```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("business_dataset_cleaned_1500_rows1.csv")

# Group by Year and sum Revenue
year_revenue = df.groupby('year')['revenue'].sum().reset_index()

# Plot line chart
plt.figure(figsize=(10, 6)) # Make the figure slightly larger for labels
plt.plot(year_revenue['year'], year_revenue['revenue'], marker='o') # Add markers for better visibility

# Add data labels
for index, row in year_revenue.iterrows():
    plt.text(row['year'], row['revenue'], f'{row['revenue']:,}', ha='center', va='bottom') # Format with commas

plt.title("Revenue Trend Over Years")
plt.xlabel("Year")
plt.ylabel("Revenue (Sum)")
plt.grid(True, linestyle='--', alpha=0.7) # Add a grid for readability
plt.xticks(year_revenue['year']) # Ensure all years are shown as ticks
plt.tight_layout() # Adjust layout to prevent labels from being cut off
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("business_dataset_cleaned_1500_rows1.csv")

# Group by Month and sum Sales Amount
month_sales = df.groupby('month')['sales_amount'].sum().reset_index()

# Sort months if needed (ensure consistent order for plotting)
# Define a custom order for months
month_order = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
month_sales['month'] = pd.Categorical(month_sales['month'], categories=month_order, ordered=True)
month_sales = month_sales.sort_values('month')

# Plot line chart
plt.figure(figsize=(12, 7)) # Make the figure larger for labels and readability
plt.plot(month_sales['month'], month_sales['sales_amount'], marker='o') # Add markers for better visibility

# Add data labels
for index, row in month_sales.iterrows():
    plt.text(row['month'], row['sales_amount'], f'{row['sales_amount']:,}', ha='center', va='bottom', fontsize=9) # Format with commas, adjust fontsize

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales Amount")
plt.grid(True, linestyle='--', alpha=0.7) # Add a grid for readability
plt.tight_layout() # Adjust layout to prevent labels from being cut off
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("business_dataset_cleaned_1500_rows1.csv")

# Group by Quarter and sum Profit
quarter_profit = df.groupby('quarter')['profit'].sum().reset_index()

# Sort quarters if needed
quarter_profit = quarter_profit.sort_values('quarter')

# Plot line chart
plt.figure(figsize=(10, 6)) # Make the figure slightly larger for labels
plt.plot(quarter_profit['quarter'], quarter_profit['profit'], marker='o') # Add markers for better visibility

# Add data labels
for index, row in quarter_profit.iterrows():
    plt.text(row['quarter'], row['profit'], f'{row['profit']:,}', ha='center', va='bottom') # Format with commas

plt.title("Quarterly Profit Trend")
plt.xlabel("Quarter")
plt.ylabel("Profit")
plt.grid(True, linestyle='--', alpha=0.7) # Add a grid for readability
plt.tight_layout() # Adjust layout to prevent labels from being cut off
plt.show()
```

```python
product_revenue = df.groupby('product_name')['revenue'].sum().reset_index()
product_revenue = product_revenue.sort_values('revenue', ascending=False)

plt.figure(figsize=(10,6))
bars = plt.bar(product_revenue['product_name'], product_revenue['revenue'])

for bar in bars:
    y = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, y, f'{y:,.0f}',
             ha='center', va='bottom')

plt.title("Revenue by Product")
plt.xlabel("Product Name")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
product_margin = df.groupby('product_name')['profit_margin'].mean().reset_index()
product_margin = product_margin.sort_values('profit_margin', ascending=False)

plt.figure(figsize=(10,6))
bars = plt.bar(product_margin['product_name'], product_margin['profit_margin'])

for bar in bars:
    y = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, y, f'{y:.2f}',
             ha='center', va='bottom')

plt.title("Profit Margin by Product")
plt.xlabel("Product Name")
plt.ylabel("Average Profit Margin")
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```
